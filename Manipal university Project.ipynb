## Arithmetic Expression Examples

In Data Science, arithmetic expressions are used for calculations and data manipulation. Below are some common examples:

- **Addition**: `5 + 3 = 8`
- **Subtraction**: `10 - 4 = 6`
- **Multiplication**: `6 * 7 = 42`
- **Division**: `20 / 5 = 4.0`
- **Exponentiation**: `2 ** 3 = 8`
- **Modulus (Remainder)**: `10 % 3 = 1`
- **Floor Division**: `17 // 3 = 5`
